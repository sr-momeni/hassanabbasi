```python
from typing import Self
#the first list would be related to the users information(include age, university, major, years of experience)
class User:
    def __init__(self, age, university, major, years_of_experience):
        self.age = age
        self.university = university
        self.major = major
        self.years_of_experience = years_of_experience

    def Approval(self):
      if self.years_of_experience > 2:
        return f'{self.university} user has been approved' # Fixed self.name since name is not an attribute
      else:
        return f'{self.university} user has not been approved'

    def age_restriction(self): # Added missing 'self' parameter
      if self.age >= 28:
        return f'{self.university} user can get this job'
      else:
        return f'{self.university} user cannot get this job'

User1 = User(25,'uOttawa', 'Computer', 2)
User2 = User(28,'Carleton University', 'Computer', 4)
User3 = User(32,'uOttawa', 'Electrical', 1)
User4 = User(24,'uOttawa', 'Mechanical', 1)

user_list = [User1, User2, User3, User4]

# Convert the list of custom objects to a list of dictionaries to load into pandas correctly
user_data = [vars(user) for user in user_list]
illustation = pd.DataFrame(user_data)
illustation
```

```python
import numpy as np

# Approach 1: Convert pandas DataFrame columns directly to NumPy arrays
ages_arr = illustation['age'].to_numpy()
universities_arr = illustation['university'].to_numpy()
majors_arr = illustation['major'].to_numpy()
experience_arr = illustation['years_of_experience'].to_numpy()

# Approach 2: Create a 2D structured or object NumPy array from the list of objects
# We can create a 2D array of all user values:
user_matrix = np.array([
    [user.age, user.university, user.major, user.years_of_experience]
    for user in user_list
], dtype=object)

# Let's print the individual 1D arrays and the 2D matrix
print("Ages Array:", ages_arr)
print("Experience Array:", experience_arr)
print("\n2D User Matrix:\n", user_matrix)
```
